## Aim

To study the matrix representation and operation of the Identity (I) gate on a single qubit using PennyLane.

## Software / Tools Required

Python 3.10+, Jupyter Notebook, PennyLane, NumPy

## Procedure

1. Import PennyLane and NumPy.
2. Define the 2 × 2 matrix of the Identity gate.
3. Create a single-qubit quantum circuit and apply the Identity gate.
4. Obtain and display the resulting state and the gate matrix.
5. Observe that the Identity gate leaves the qubit state unchanged.

## Code



```python
import pennylane as qml
import numpy as np

I = qml.Identity(0).matrix()
print("Identity gate matrix:")
print(I)

dev = qml.device("default.qubit", wires=1)

@qml.qnode(dev)
def circuit():
    qml.PauliX(0)
    qml.Identity(0)
    return qml.state()

state = circuit()
print("\nState after X followed by Identity:")
print(state)
```

## Output / Observation

Run the code cells above to generate the actual output. The expected observation is:

The Identity gate has matrix [[1, 0], [0, 1]] and does not change the quantum state.
