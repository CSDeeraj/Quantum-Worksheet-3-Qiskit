## Aim

To study the matrix representation, operation, and simulation of the Pauli-X gate on a single qubit using PennyLane.

## Software / Tools Required

Python 3.10+, Jupyter Notebook, PennyLane, NumPy

## Procedure

1. Import PennyLane and NumPy.
2. Define the Pauli-X gate matrix.
3. Initialize a qubit in |0⟩.
4. Apply the Pauli-X gate.
5. Display the resulting state and verify that |0⟩ changes to |1⟩.

## Code



```python
import pennylane as qml
import numpy as np

X = qml.PauliX(0).matrix()
print("Pauli-X gate matrix:")
print(X)

dev = qml.device("default.qubit", wires=1)

@qml.qnode(dev)
def circuit():
    qml.PauliX(0)
    return qml.state()

state = circuit()
print("\nState after applying X to |0>:")
print(state)
print("\nProbabilities:")
print(np.abs(state) ** 2)
```

## Output / Observation

Run the code cells above to generate the actual output. The expected observation is:

The Pauli-X matrix is [[0, 1], [1, 0]]. Applying it to |0⟩ produces |1⟩.

## Result

The Pauli-X gate was successfully simulated and its bit-flip action was verified.
