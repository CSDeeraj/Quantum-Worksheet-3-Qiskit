Worksheet 3 — Matrix Representation, Verification, and Simulation of Fundamental Single-Qubit Quantum Gates using PennyLane.

```python
import numpy as np
import pennylane as qml
from pennylane import numpy as pnp

np.set_printoptions(precision=4, suppress=True)

def verify_matrix(name, U):
    print(f"{name} matrix:")
    print(U)
    print("\nU†U =")
    print(U.conj().T @ U)
    print("Unitary:", np.allclose(U.conj().T @ U, np.eye(2)))
```

```python
X = np.array([[0,1],[1,0]], dtype=complex)
print("X|0> =", X @ np.array([1,0], complex))
print("X|1> =", X @ np.array([0,1], complex))

dev = qml.device("default.qubit", wires=1)
@qml.qnode(dev)
def x_qnode():
    qml.PauliX(0)
    return qml.state()
print("PennyLane X|0> ->", x_qnode())
```
