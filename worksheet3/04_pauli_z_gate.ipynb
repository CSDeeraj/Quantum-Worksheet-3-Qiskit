## Aim

To study the matrix representation, operation, and simulation of the Pauli-Z gate on a single qubit using PennyLane.

## Software / Tools Required

Python 3.10+, Jupyter Notebook, PennyLane, NumPy

## Procedure

1. Import PennyLane and NumPy.
2. Obtain the matrix representation of the Pauli-Z gate.
3. Prepare the qubit in |1⟩ so that the phase-flip action is visible.
4. Apply the Pauli-Z gate.
5. Display the resulting state and probabilities.

## Code



```python
import pennylane as qml
import numpy as np

Z = qml.PauliZ(0).matrix()
print("Pauli-Z gate matrix:")
print(Z)

dev = qml.device("default.qubit", wires=1)

@qml.qnode(dev)
def circuit():
    qml.PauliX(0)  # Prepare |1>
    qml.PauliZ(0)
    return qml.state()

state = circuit()
print("\nState after preparing |1> and applying Z:")
print(state)
print("\nProbabilities:")
print(np.abs(state) ** 2)
```

## Output / Observation

Run the code cells above to generate the actual output. The expected observation is:

The Pauli-Z matrix is [[1, 0], [0, -1]]. It leaves |0⟩ unchanged and changes the phase of |1⟩.

## Result

The Pauli-Z gate was successfully simulated and its phase-flip behaviour was verified.
