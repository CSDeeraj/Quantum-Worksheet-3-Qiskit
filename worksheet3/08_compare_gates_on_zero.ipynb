## Aim

To compare the action of fundamental single-qubit quantum gates on the |0⟩ state using PennyLane.

## Software / Tools Required

Python 3.10+, Jupyter Notebook, PennyLane, NumPy

## Procedure

1. Define the Identity, Pauli-X, Pauli-Y, Pauli-Z, S, T, and Hadamard gates.
2. Apply each gate separately to the |0⟩ state.
3. Calculate the resulting state vectors.
4. Calculate measurement probabilities.
5. Compare the outputs and identify bit-flip, phase-shift, and superposition behaviour.

## Code



```python
import pennylane as qml
import numpy as np

gates = {
    "Identity": qml.Identity,
    "Pauli-X": qml.PauliX,
    "Pauli-Y": qml.PauliY,
    "Pauli-Z": qml.PauliZ,
    "S": qml.S,
    "T": qml.T,
    "Hadamard": qml.Hadamard,
}

dev = qml.device("default.qubit", wires=1)

def apply_gate(gate):
    @qml.qnode(dev)
    def circuit():
        gate(0)
        return qml.state()
    return circuit()

for name, gate in gates.items():
    state = apply_gate(gate)
    probs = np.abs(state) ** 2
    print(f"\n{name}")
    print("State:", state)
    print("Probabilities:", probs)
```

## Output / Observation

Run the code cells above to generate the actual output. The expected observation is:

X and Y transfer |0⟩ to the |1⟩ basis state (with a phase difference for Y). Z, S and T do not change |0⟩ except for a global phase of 1. H creates an equal superposition.

## Result

The fundamental single-qubit gates were successfully compared on the |0⟩ input state.
