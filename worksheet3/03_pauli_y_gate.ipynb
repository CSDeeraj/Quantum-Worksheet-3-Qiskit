## Aim

To study the matrix representation, operation, and simulation of the Pauli-Y gate on a single qubit using PennyLane.

## Software / Tools Required

Python 3.10+, Jupyter Notebook, PennyLane, NumPy

## Procedure

1. Import PennyLane and NumPy.
2. Obtain the matrix representation of the Pauli-Y gate.
3. Initialize a qubit in |0⟩.
4. Apply the Pauli-Y gate.
5. Display the resulting state and probabilities.

## Code



```python
import pennylane as qml
import numpy as np

Y = qml.PauliY(0).matrix()
print("Pauli-Y gate matrix:")
print(Y)

dev = qml.device("default.qubit", wires=1)

@qml.qnode(dev)
def circuit():
    qml.PauliY(0)
    return qml.state()

state = circuit()
print("\nState after applying Y to |0>:")
print(state)
print("\nProbabilities:")
print(np.abs(state) ** 2)
```

## Output / Observation

Run the code cells above to generate the actual output. The expected observation is:

The Pauli-Y gate introduces both a bit flip and a phase factor. For |0⟩, the resulting state is i|1⟩ up to the convention used by the simulator.
