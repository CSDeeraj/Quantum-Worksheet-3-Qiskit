# Program 3 — Pauli-Y gate

Worksheet 3 — Matrix Representation, Verification, and Simulation of Fundamental Single-Qubit Quantum Gates using PennyLane.

```python
import numpy as np
import pennylane as qml
from pennylane import numpy as pnp

np.set_printoptions(precision=4, suppress=True)

def verify_matrix(name, U):
    print(f"{name} matrix:")
    print(U)
    print("\nU†U =")
    print(U.conj().T @ U)
    print("Unitary:", np.allclose(U.conj().T @ U, np.eye(2)))
```

```python
Y = np.array([[0,-1j],[1j,0]], dtype=complex)
print("Y|0> =", Y @ np.array([1,0], complex))
print("Y|1> =", Y @ np.array([0,1], complex))

dev = qml.device("default.qubit", wires=1)
@qml.qnode(dev)
def y_qnode():
    qml.PauliY(0)
    return qml.state()
print("PennyLane Y|0> ->", y_qnode())
```
