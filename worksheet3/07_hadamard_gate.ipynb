## THEORY

PennyLane is an open-source quantum computing software framework developed by Xanadu for designing, simulating, and optimizing quantum circuits. A quantum device represents either a quantum simulator or real quantum hardware on which quantum circuits are executed. In PennyLane, devices are created using `qml.device()`.

A Quantum Node (QNode) is the fundamental computational unit in PennyLane. It combines a quantum circuit with a quantum device, allowing quantum operations to be executed and measurement results to be returned to classical programs. QNodes are created using the `@qml.qnode` decorator.

Quantum gates are unitary operations that manipulate the state of qubits. PennyLane provides gates such as Pauli-X, Pauli-Y, Pauli-Z, Hadamard, Phase (S), and T gates. Measurements can return state vectors, probabilities, expectation values, and samples.

# PROGRAM 7

**Objective:** Implement the Hadamard gate matrix and verify the creation of the equal superposition state using analytical calculations and PennyLane simulation.


```python
import numpy as np
import pennylane as qml

H = (1 / np.sqrt(2)) * np.array([[1, 1], [1, -1]], dtype=complex)
zero = np.array([1, 0], dtype=complex)

analytical_state = H @ zero

print("Hadamard matrix:")
print(H)
print("\nH|0> =", analytical_state)
print("Analytical probabilities =", np.abs(analytical_state) ** 2)

dev = qml.device("default.qubit", wires=1)

@qml.qnode(dev)
def h_qnode():
    qml.Hadamard(0)
    return qml.state()

simulated_state = h_qnode()
print("\nPennyLane state =", simulated_state)
print("PennyLane probabilities =", np.abs(simulated_state) ** 2)
```

## OUTPUT

The output is generated by the code cell above when the notebook is executed.


## RESULT

The Hadamard gate transforms |0⟩ into an equal superposition of |0⟩ and |1⟩, with probabilities 1/2 and 1/2. Analytical and PennyLane results agree.
