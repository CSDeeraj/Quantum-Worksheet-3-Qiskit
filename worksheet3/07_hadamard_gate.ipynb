## Aim

To study the matrix representation and superposition-generating property of the Hadamard gate using PennyLane.

## Software / Tools Required

Python 3.10+, Jupyter Notebook, PennyLane, NumPy

## Procedure

1. Import PennyLane and NumPy.
2. Obtain the Hadamard matrix.
3. Initialize a qubit in |0⟩.
4. Apply the Hadamard gate.
5. Display the resulting state and probabilities to verify equal superposition.

## Code



```python
import pennylane as qml
import numpy as np

H = qml.Hadamard(0).matrix()
print("Hadamard gate matrix:")
print(H)

dev = qml.device("default.qubit", wires=1)

@qml.qnode(dev)
def circuit():
    qml.Hadamard(0)
    return qml.state()

state = circuit()
print("\nState after applying H to |0>:")
print(state)
print("\nProbabilities:")
print(np.abs(state) ** 2)
```

## Output / Observation

Run the code cells above to generate the actual output. The expected observation is:

Applying H to |0⟩ produces (|0⟩ + |1⟩)/√2, giving approximately 0.5 probability for each computational basis state.

## Result

The Hadamard gate was successfully simulated and its superposition property was verified.
