Worksheet 3 — Matrix Representation, Verification, and Simulation of Fundamental Single-Qubit Quantum Gates using PennyLane.

```python
import numpy as np
import pennylane as qml
from pennylane import numpy as pnp

np.set_printoptions(precision=4, suppress=True)

def verify_matrix(name, U):
    print(f"{name} matrix:")
    print(U)
    print("\nU†U =")
    print(U.conj().T @ U)
    print("Unitary:", np.allclose(U.conj().T @ U, np.eye(2)))
```

```python
H = (1/np.sqrt(2))*np.array([[1,1],[1,-1]], dtype=complex)
zero = np.array([1,0], complex)
print("H|0> =", H @ zero)
print("Expected probabilities =", np.abs(H @ zero)**2)

dev = qml.device("default.qubit", wires=1)
@qml.qnode(dev)
def h_qnode():
    qml.Hadamard(0)
    return qml.state()
print("PennyLane H|0> ->", h_qnode())
print("PennyLane probabilities ->", np.abs(h_qnode())**2)
```
