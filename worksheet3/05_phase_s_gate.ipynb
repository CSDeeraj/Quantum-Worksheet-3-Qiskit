## Aim

To study the matrix representation and phase-shift operation of the S gate on a single qubit using PennyLane.

## Software / Tools Required

Python 3.10+, Jupyter Notebook, PennyLane, NumPy

## Procedure

1. Import PennyLane and NumPy.
2. Obtain the matrix representation of the S gate.
3. Prepare |1⟩ to make the phase change observable.
4. Apply the S gate.
5. Display the resulting state and verify the phase factor i.

## Code



```python
import pennylane as qml
import numpy as np

S = qml.S(0).matrix()
print("S gate matrix:")
print(S)

dev = qml.device("default.qubit", wires=1)

@qml.qnode(dev)
def circuit():
    qml.PauliX(0)  # Prepare |1>
    qml.S(0)
    return qml.state()

state = circuit()
print("\nState after preparing |1> and applying S:")
print(state)
print("\nProbabilities:")
print(np.abs(state) ** 2)
```

## Output / Observation

Run the code cells above to generate the actual output. The expected observation is:

The S gate matrix is diagonal with entries 1 and i. It changes the phase of |1⟩ without changing measurement probabilities.

## Result

The S phase gate was successfully represented and its phase-shift action was verified.
