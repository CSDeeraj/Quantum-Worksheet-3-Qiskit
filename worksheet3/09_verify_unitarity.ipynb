# Program 9 — Verify unitary property

```python
import numpy as np
import pennylane as qml
from pennylane import numpy as pnp

np.set_printoptions(precision=4, suppress=True)

def verify_matrix(name, U):
    print(f"{name} matrix:")
    print(U)
    print("\nU†U =")
    print(U.conj().T @ U)
    print("Unitary:", np.allclose(U.conj().T @ U, np.eye(2)))

gate_matrices = {
    "Pauli-X": np.array([[0,1],[1,0]], complex),
    "Pauli-Y": np.array([[0,-1j],[1j,0]], complex),
    "Pauli-Z": np.array([[1,0],[0,-1]], complex),
    "Phase (S)": np.array([[1,0],[0,1j]], complex),
    "T": np.array([[1,0],[0,np.exp(1j*np.pi/4)]], complex),
    "Hadamard": (1/np.sqrt(2))*np.array([[1,1],[1,-1]], complex)
}

for name, U in gate_matrices.items():
    product = U.conj().T @ U
    print(f"{name}:")
    print(product)
    print("Unitary =", np.allclose(product, np.eye(2)))
    print()
```
