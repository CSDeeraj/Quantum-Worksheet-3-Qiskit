## Aim

To verify mathematically that the fundamental single-qubit quantum gate matrices are unitary using PennyLane and NumPy.

## Software / Tools Required

Python 3.10+, Jupyter Notebook, PennyLane, NumPy

## Procedure

1. Obtain the matrix representation of each fundamental gate.
2. For each matrix U, calculate its conjugate transpose U†.
3. Compute U†U.
4. Compare U†U with the identity matrix using NumPy.
5. Report whether each gate satisfies the unitarity condition.

## Code



```python
import pennylane as qml
import numpy as np

gates = {
    "Identity": qml.Identity(0).matrix(),
    "Pauli-X": qml.PauliX(0).matrix(),
    "Pauli-Y": qml.PauliY(0).matrix(),
    "Pauli-Z": qml.PauliZ(0).matrix(),
    "S": qml.S(0).matrix(),
    "T": qml.T(0).matrix(),
    "Hadamard": qml.Hadamard(0).matrix(),
}

identity = np.eye(2)

for name, U in gates.items():
    product = U.conj().T @ U
    unitary = np.allclose(product, identity)

    print(f"\n{name}")
    print("U†U =")
    print(product)
    print("Unitary:", unitary)
```

Run the code cells above to generate the actual output. The expected observation is:

The unitarity of the fundamental single-qubit quantum gates was successfully verified.
