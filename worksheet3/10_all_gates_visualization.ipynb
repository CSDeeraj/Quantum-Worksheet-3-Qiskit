# Program 10 — Fundamental one-qubit gates: matrices, simulation, and probability distributions

```python
import numpy as np
import pennylane as qml
from pennylane import numpy as pnp

np.set_printoptions(precision=4, suppress=True)

def verify_matrix(name, U):
    print(f"{name} matrix:")
    print(U)
    print("\nU†U =")
    print(U.conj().T @ U)
    print("Unitary:", np.allclose(U.conj().T @ U, np.eye(2)))

import matplotlib.pyplot as plt

gate_matrices = {
    "I": np.array([[1,0],[0,1]], complex),
    "X": np.array([[0,1],[1,0]], complex),
    "Y": np.array([[0,-1j],[1j,0]], complex),
    "Z": np.array([[1,0],[0,-1]], complex),
    "S": np.array([[1,0],[0,1j]], complex),
    "T": np.array([[1,0],[0,np.exp(1j*np.pi/4)]], complex),
    "H": (1/np.sqrt(2))*np.array([[1,1],[1,-1]], complex)
}

zero = np.array([1,0], complex)

fig, ax = plt.subplots(figsize=(10,5))
names = list(gate_matrices)
x = np.arange(len(names))
width = 0.35

p0 = []
p1 = []
for name in names:
    state = gate_matrices[name] @ zero
    probs = np.abs(state)**2
    p0.append(probs[0])
    p1.append(probs[1])

ax.bar(x - width/2, p0, width, label="P(0)")
ax.bar(x + width/2, p1, width, label="P(1)")
ax.set_xticks(x)
ax.set_xticklabels(names)
ax.set_ylim(0, 1.05)
ax.set_ylabel("Probability")
ax.set_title("Probability distributions after applying each gate to |0⟩")
ax.legend()
plt.show()

print("Matrix/state verification:")
for name, U in gate_matrices.items():
    state = U @ zero
    print(f"{name}: state={state}, probabilities={np.abs(state)**2}")
```
