## THEORY

PennyLane is an open-source quantum computing software framework developed by Xanadu for designing, simulating, and optimizing quantum circuits. A quantum device represents either a quantum simulator or real quantum hardware on which quantum circuits are executed. In PennyLane, devices are created using `qml.device()`.

A Quantum Node (QNode) is the fundamental computational unit in PennyLane. It combines a quantum circuit with a quantum device, allowing quantum operations to be executed and measurement results to be returned to classical programs. QNodes are created using the `@qml.qnode` decorator.

Quantum gates are unitary operations that manipulate the state of qubits. PennyLane provides gates such as Pauli-X, Pauli-Y, Pauli-Z, Hadamard, Phase (S), and T gates. Measurements can return state vectors, probabilities, expectation values, and samples.

**Objective:** Develop a PennyLane program to implement all the fundamental one-qubit quantum gates, compare their matrix representations with simulated quantum state transformations, and visualize the resulting probability distributions.


```python
import numpy as np
import pennylane as qml
import matplotlib.pyplot as plt

gate_matrices = {
    "I": np.array([[1, 0], [0, 1]], complex),
    "X": np.array([[0, 1], [1, 0]], complex),
    "Y": np.array([[0, -1j], [1j, 0]], complex),
    "Z": np.array([[1, 0], [0, -1]], complex),
    "S": np.array([[1, 0], [0, 1j]], complex),
    "T": np.array([[1, 0], [0, np.exp(1j * np.pi / 4)]], complex),
    "H": (1 / np.sqrt(2)) * np.array([[1, 1], [1, -1]], complex)
}

zero = np.array([1, 0], complex)

print("Matrix and state transformation comparison:")
probabilities = {}

for name, U in gate_matrices.items():
    state = U @ zero
    probabilities[name] = np.abs(state) ** 2
    print(f"{name}:")
    print("Matrix =")
    print(U)
    print("State = ", state)
    print("Probabilities = ", probabilities[name])
    print()

names = list(gate_matrices.keys())
x = np.arange(len(names))
width = 0.35

p0 = [probabilities[name][0] for name in names]
p1 = [probabilities[name][1] for name in names]

plt.figure(figsize=(10, 5))
plt.bar(x - width / 2, p0, width, label="P(0)")
plt.bar(x + width / 2, p1, width, label="P(1)")
plt.xticks(x, names)
plt.ylim(0, 1.05)
plt.ylabel("Probability")
plt.xlabel("Gate")
plt.title("Probability distributions after applying gates to |0⟩")
plt.legend()
plt.show()
```

## OUTPUT

The output is generated by the code cell above. It displays the matrix representations, transformed states, probabilities, and the resulting probability-distribution visualization.

## RESULT

All fundamental one-qubit gates are implemented and their matrix-based transformations are compared with the resulting probability distributions. The visualization demonstrates how each gate changes the computational-basis probabilities.
