## Aim

To visualize and compare the matrix representations of fundamental single-qubit quantum gates using PennyLane, NumPy, and Matplotlib.

## Software / Tools Required

Python 3.10+, Jupyter Notebook, PennyLane, NumPy, Matplotlib

## Procedure

1. Import PennyLane, NumPy, and Matplotlib.
2. Obtain the matrix representation of the fundamental gates.
3. Display the real and imaginary components of each matrix as heatmaps.
4. Add labels for computational basis states.
5. Compare the matrices visually and relate their structures to their quantum operations.

## Code



```python
import pennylane as qml
import numpy as np
import matplotlib.pyplot as plt

gates = {
    "Identity": qml.Identity(0).matrix(),
    "Pauli-X": qml.PauliX(0).matrix(),
    "Pauli-Y": qml.PauliY(0).matrix(),
    "Pauli-Z": qml.PauliZ(0).matrix(),
    "S": qml.S(0).matrix(),
    "T": qml.T(0).matrix(),
    "Hadamard": qml.Hadamard(0).matrix(),
}

for name, matrix in gates.items():
    fig, ax = plt.subplots(figsize=(4, 3))
    ax.imshow(np.abs(matrix), interpolation="nearest")
    ax.set_title(f"{name} gate |matrix|")
    ax.set_xticks([0, 1], labels=["|0⟩", "|1⟩"])
    ax.set_yticks([0, 1], labels=["|0⟩", "|1⟩"])
    for i in range(2):
        for j in range(2):
            ax.text(j, i, f"{matrix[i, j]:.2g}", ha="center", va="center")
    plt.tight_layout()
    plt.show()
```
