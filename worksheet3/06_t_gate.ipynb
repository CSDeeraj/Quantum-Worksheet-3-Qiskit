## Aim

To study the matrix representation and phase-shift operation of the T gate on a single qubit using PennyLane.

## Software / Tools Required

Python 3.10+, Jupyter Notebook, PennyLane, NumPy

## Procedure

1. Import PennyLane and NumPy.
2. Obtain the matrix representation of the T gate.
3. Prepare the qubit in |1⟩.
4. Apply the T gate.
5. Display the resulting state and verify the phase factor exp(iπ/4).

## Code



```python
import pennylane as qml
import numpy as np

T = qml.T(0).matrix()
print("T gate matrix:")
print(T)

dev = qml.device("default.qubit", wires=1)

@qml.qnode(dev)
def circuit():
    qml.PauliX(0)  # Prepare |1>
    qml.T(0)
    return qml.state()

state = circuit()
print("\nState after preparing |1> and applying T:")
print(state)
print("\nProbabilities:")
print(np.abs(state) ** 2)
```

## Output / Observation

Run the code cells above to generate the actual output. The expected observation is:

The T gate applies a π/4 phase shift to the |1⟩ component while preserving measurement probabilities.
